```python
class Celular:
    # Constructor
    def __init__(self, marca, modelo, precio):
        self.marca = marca
        self.modelo = modelo
        self.precio = precio
        self.encendido = False

    # Métodos
    def encender(self):
        self.encendido = True
        return f"El {self.modelo} está encendido."

    def mostrar_info(self):
        return f"Celular: {self.marca} {self.modelo}, Precio: ${self.precio}"

# Crear instancias (objetos)
celular1 = Celular("Samsung", "Galaxy S23", 800)
celular2 = Celular("Apple", "iPhone 15", 1000)
# llamadas a las funciones para cada objeto celular 1 y celular 2
print("celular1")
print(celular1.mostrar_info())
print(celular1.encender())
print("celular2")
print(celular2.mostrar_info())
print(celular2.encender())



print(celular1.mostrar_info())
print(celular2.encender())

print("creador del software por yahir sifuentes NL = 56 ")
```
